```python
import pymysql

# MySQL 연결 설정
connection = pymysql.connect(
    host= '192.168.2.131',  # 도커 컨테이너가 로컬에서 실행 중이므로 localhost
    port=3307,         # 도커 실행 시 매핑한 포트 (3307:3306)
    user='root',       # MySQL 사용자 이름
    password='1234',   # 도커 실행 시 설정한 루트 비밀번호
    database='mydb'   # 기본 데이터베이스 (없으면 생성한 데이터베이스 이름)
)

try:
    with connection.cursor() as cursor:
        # 1) 테이블 생성 (없으면 생성)
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS employees (
                id INT AUTO_INCREMENT PRIMARY KEY,
                name VARCHAR(100),
                position VARCHAR(100)
            )
        """)
        connection.commit()

        # 2) 데이터 삽입 (예시)
        cursor.execute("INSERT INTO employees (name, position) VALUES (%s, %s)", ("Alice", "Developer"))
        cursor.execute("INSERT INTO employees (name, position) VALUES (%s, %s)", ("Bob", "Designer"))
        connection.commit()

        # 3) 데이터 조회
        cursor.execute("SELECT * FROM employees")
        rows = cursor.fetchall()

        print("employees 테이블 데이터:")
        for row in rows:
            print(row)

finally:
    connection.close()
```
