```python
import folium
import requests
from pandas import json_normalize
```

```python
# 서울 중심 좌표
seoul_coords = [37.5662952, 126.9779451]
# 지도 생성
map = folium.Map(location=seoul_coords, zoom_start=15)

# 마커 추가
folium.Marker(
    location=seoul_coords,
    popup="서울특별시",  # 마커 클릭 시 표시될 글
    tooltip="서울"       # 마우스 오버 시 표시될 글
).add_to(map)


map
```

```python
targetSite='https://www.starbucks.co.kr/store/getStore.do?r=8K3XF06R9Q'
req=requests.post(targetSite, data={
'ins_lat': 37.563398,
'ins_lng': 126.9863309,
'p_sido_cd': '01',
'p_gugun_cd': '',
'in_biz_cd': '',
'iend': 600,
'set_date': ''
})
starbucks=req.json()
print(type(starbucks))
```

```python
st_df = json_normalize(starbucks, 'list')
st_df
```

```python
st_df_map = st_df[['s_name','sido_code','sido_name','gugun_code','gugun_name',
                   'doro_address','lat','lot']]
st_df_map.info()
```

```python
st_df_map['lat'] = st_df_map['lat'].astype(float)
st_df_map['lot'] = st_df_map['lot'].astype(float)
```

```python
st_df_map.info()
```

```python
st_df_map.head()
```

```python
loc = [37.5662952, 126.9779451]
Map = folium.Map(location=loc, zoom_start=18)
pop = folium.Popup('역삼아레나빌딩', max_width=100)
folium.Marker(location=loc, popup=pop).add_to(Map)
Map
```

```python
map = folium.Map(location=[37.5662952, 126.9779451], zoom_start=10)

# 앞 5개 매장만 선택
for idx, row in st_df_map.head(5).iterrows():
    folium.Marker(
        location=[row['lat'], row['lot']],
        popup=row['s_name'],  # 클릭 시 매장 이름 표시
        tooltip=row['sido_name'] + ' ' + row['gugun_name'],  # 마우스 오버 시 시도+구 표시
    ).add_to(map)
map
```

```python
map = folium.Map(location=[37.5662952, 126.9779451], zoom_start=10)

# 모든 스타벅스 매장
for idx, row in st_df_map.iterrows():
    folium.Marker(
        location=[row['lat'], row['lot']]
    ).add_to(map)
map
```
