```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# 단순선형회귀 모델 만들기
```

```python
X = np.random.rand(100)
y = 0.2 * X * 0.5 
```

```python
def plt_prediction(pred, y):
    plt.figure(figsize=(8,6))
    plt.scatter(X, y, color='blue', label='Actual')
    plt.scatter(X, pred, color='orange', label='Prediction')
    plt.legend()
    plt.show()
```
