```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))

try:
    url = 'https://www.nike.com/kr/w/new-releases-shoes-3n82yzy7ok'
    driver.get(url)
    time.sleep(5)

    # 스크롤 끝까지 내리기
    last_height = driver.execute_script("return document.body.scrollHeight")
    while True:
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        new_height = driver.execute_script("return document.body.scrollHeight")
        if new_height == last_height:
            break
        last_height = new_height

        # 상품명
    names = driver.find_elements(By.CSS_SELECTOR, 'div.product-card__titles > div.product-card__title')
    # 신발 종류
    types = driver.find_elements(By.CSS_SELECTOR, 'div.product-card__titles > div.product-card__subtitle')
    # 가격
    prices = driver.find_elements(By.CSS_SELECTOR, 'div.product-price__wrapper > div.product-price')
    # 이미지 URL
    images = driver.find_elements(By.CSS_SELECTOR, 'div.wall-image-loader > img')

    print(f"상품 개수: {len(names)}개")
    data = []
    for i in range(len(names)):
        name = names[i].text
        price = prices[i].text if i < len(prices) else "가격 정보 없음"
        shoe_type = types[i].text if i < len(types) else "종류 정보 없음"
        img_url = images[i].get_attribute('src') if i < len(images) else "이미지 없음"
        data.append(( name, price, shoe_type,img_url))


finally:
    driver.quit()
```

```python
# sqlight 생성 저장
import sqlite3

# DB 연결 및 테이블 생성
sqlite_conn = sqlite3.connect("nike_shoes.db")  # 로컬에 DB 파일 생성
sqlite_cursor = sqlite_conn.cursor()

sqlite_cursor.execute('''
CREATE TABLE IF NOT EXISTS shoes (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    category TEXT,
    name TEXT,
    price TEXT,
    image_url TEXT
)
''')

# 데이터 삽입
sqlite_cursor.executemany('REPLACE INTO shoes (name, price, category, image_url) VALUES (?, ?, ?, ?)', data)
sqlite_conn.commit()
print("데이터 삽입 완료!")
```
