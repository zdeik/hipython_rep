```python
from sklearn.model_selection import KFold
KFold = KFold(n_splits=5)
iris_clf = DecisionTreeClassifier(random_state=156)
iris_data, iris_label
```

```python
# KFold를 이용한 5-Fold 교차검증 수행
for train_index, test_index in KFold.split(iris_data):
    # 훈련 데이터와 테스트 데이터로 분할 (인덱스를 기준으로 슬라이싱)
    X_train, X_test = iris_data[train_index], iris_data[test_index]
    y_train, y_test = iris_label[train_index], iris_label[test_index]
    
    # 훈련 데이터를 기반으로 모델 학습
    iris_clf.fit(X_train, y_train)
    
    # 테스트 데이터에 대한 예측 수행
    pred = iris_clf.predict(X_test)
    
    # 예측 결과와 실제 라벨을 비교하여 정확도(Accuracy) 계산
    accuracy = accuracy_score(y_test, pred)
    
    # 현재 Fold에 대한 정확도 출력
    print(f'Accuracy: {accuracy:.4f}')
```

# 알고리즘에 따른 스케일링의 효과 확인

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=11)  # stratify=y
```

```python
# 스케일링 없이 알고리즘에 따른 성능
models = {
    'KNN': KNeighborsClassifier(n_neighbors=5),
    'RF' : RandomForestClassifier(),
    'LR' : LogisticRegression(max_iter=3000)
}
for name, model in models.items():
    model.fit(X_train, y_train)
    pred_models = model.predict(X_test)
    acc = accuracy_score(y_test, pred_models)
    print(f'{name}모델의 정확도: {acc:.5f}')
    
    
```

```python
# 알고리즘에 따른 스케일링 성능 변화
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

models = {
    'KNN': KNeighborsClassifier(n_neighbors=5),
    'RF' : RandomForestClassifier(),
    'LR' : LogisticRegression(max_iter=3000)
}
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    pred_models = model.predict(X_test_scaled)
    acc = accuracy_score(y_test, pred_models)
    print(f'{name}모델의 정확도: {acc:.5f}')    
```
